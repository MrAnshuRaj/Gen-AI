## Agent With Memory

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
```

```python
agent = create_agent(
    model = "gpt-5",
    checkpointer=InMemorySaver()
)
```

```python
### Run with thread id
config = {"configurable":{"thread_id":"test1"}}
```

```python
from langchain_core.messages import HumanMessage,SystemMessage
response = agent.invoke({"messages":[HumanMessage(content="Hi my name is Anshu")]}, config = config)
response
```

```python
agent.invoke({"messages":[HumanMessage(content="Hi what is my name?")]},config=config)
```

```python
### Using different memory by changing thread id
config2 = {"configurable":{"thread_id":"test2"}}
```

```python
response = agent.invoke({"messages":[HumanMessage(content="Hi my name is Piyush")]}, config = config2)
response
```

```python
agent.invoke({"messages":[HumanMessage(content="Hi what is my name?")]},config=config2)
```
